```python
import pandas as pd
import matplotlib.pyplot as plt

df['date'] = pd.to_datetime(df['date'])

counts = df.groupby('date')['source_url'].count().sort_index()

plt.figure(figsize=(16, 6))
plt.bar(counts.index, counts.values)

plt.xlabel("Date")
plt.ylabel("Count")
plt.title("Observation Counts per Date")

plt.xticks(rotation=45)   # rotate dates so they are readable
plt.tight_layout()
plt.show()
```
